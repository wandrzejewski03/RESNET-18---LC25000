```python
import os
import shutil
from sklearn.model_selection import train_test_split

original_root = "/content/lung_colon_image_set"
target_root = "/content/drive/MyDrive/DL_PROJECT/lung_colon_image_set_split" #new folder
classes = ['colon_aca', 'colon_n', 'lung_aca', 'lung_n', 'lung_scc']
```

```python
for split in ['train', 'val']:
    for cls in classes:
        os.makedirs(os.path.join(target_root, split, cls), exist_ok=True) #just for creating folders in target root
```

```python
for cls in classes:
    if 'colon' in cls:
        source_dir = os.path.join(original_root, 'colon_image_sets', cls)
    else:
        source_dir = os.path.join(original_root, 'lung_image_sets', cls)

    all_images = os.listdir(source_dir)
    train_imgs, val_imgs = train_test_split(all_images, test_size=0.2, random_state=42)

    for img in train_imgs:
        shutil.copy(os.path.join(source_dir, img),
                    os.path.join(target_root, 'train', cls, img))

    for img in val_imgs:
        shutil.copy(os.path.join(source_dir, img),
                    os.path.join(target_root, 'val', cls, img))
```
